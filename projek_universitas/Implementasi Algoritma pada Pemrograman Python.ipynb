Kemudian, bandingkan ketiga metode tersebut untuk jumlah partisi $(N)$ $10$, $20$,  dan $50$. Buatlah 2 tabel dimana 1 tabel berisi nilai dari masing-masing integral dan 1 tabel lagi berisi mutlak selisih antara Metode Riemann dan Metode Eksak

**Metode Eksak**

```python
import sympy as sp
from math import *

#Definisi fungsi
x = sp.Symbol('x')
f = (1/sp.sqrt(2*pi))*sp.exp(-(x**2)/2)

#Batas integral [-3,3]
a = -3
b = 3

hasil = sp.integrate(f,(x,a,b))
eksak = sp.N(hasil) #sp.N() : Mengubah simbolik menjadi numerik
print("Hasil Metode Eksak :",eksak)
```

Dalam program ini, akan dihitung integral tentu dengan batas yang diberikan menggunakan 4 metode yaitu Metode Eksak, Metode Left Riemann Sum, Metode Middle Riemann Sum, dan Metode Right Riemann Sum. <br><br>
Metode Eksak adalah metode yang menghasilkan penyelesaian paling optimal karena metode ini pembuktiannya dilakukan secara analitis menggunakan metode matematis. Pada soal ini, digunakan Metode Eksak untuk menghitung integral tentu pada interval $a$ sampai interval $b$. <br><br>
Metode Riemann Sum atau Jumlahan Riemann adalah suatu metode pendekatan integral yang menggunakan metode penjumlahan terbatas. Metode ini menghasilkan penyelesaian dengan membagi daerah yang ingin dihitung menjadi beberapa keping dengan bentuk tertentu. Pada Metode Left, Middle, dan Right Riemann Sum, bentuk yang digunakan untuk aproksimasi fungsi yaitu beberapa persegi panjang. <br>
- Pada Left Riemann Sum, aproksimasi fungsi dihitung menggunakan nilai di titik ujung kiri dari beberapa persegi panjang dengan alas $\Delta x$ dan tinggi $f(a + i\Delta x)$. Dengan $i=0,1,2,...,n-1$, area yang dihasilkan
$$A_{kiri}=\Delta x [f(a)+f(a+\Delta x)+...+f(b-\Delta x)].$$
- Pada Middle Riemann Sum, aproksimasi fungsi dihitung menggunakan titik tengah interval dengan $f(a + \frac{\Delta x}{2})$ untuk interval pertama, $f(a + \frac{3\Delta x}{2})$ untuk interval berikutnya, dan seterusnya hingga $f(b - \frac{\Delta x}{2})$. Area yang dihasilkan
$$A_{tengah}=\Delta x [f(a+\frac{\Delta x}{2})+f(a+\frac{3\Delta x}{2})+...+f(b-\frac{\Delta x}{2})].$$
- Pada Right Riemann Sum, aproksimasi fungsi dihitung menggunakan nilai pada titik ujung kanan dari beberapa persegi panjang yang memiliki alas $Δx$ dan tinggi $f(a + i\Delta x)$. Area yang dihasilkan
$$A_{kanan}=\Delta x[f(a+\Delta x)+f(a+2\Delta x)+...+f(b)].$$

Membuat program yang meminta input suatu bilangan asli $n$ dan memberi output semua bilangan prima yang ada di $n$.

```python
n = str(input("Masukkan suatu bilangan= "))
num = []
prima = []

#Mengecek apakah n bilangan prima
def prime(n):
    if n <= 0 or n==1:
        return "Not prime"
    else:
        for i in range(2,n):
            if n%i == 0:
                return "Not prime"
        return "prime"

#Mencari seluruh kombinasi nilai dari n
for k in range(1, len(n)+1):
    for i in range(len(n)-k+1):
        num.append(int(n[i : i + k]))

num = list(set(num))

#Mengecek bilangan prima dari num
for j in num:
    x = prime(j)
    if x == "prime":
        prima.append(j)

print(prima)
```

Dalam program diberikan variabel `n` untuk input memasukkan bilangan asli, kemudian terdapat variabel `num` dan `prima` yang diinisiasikan sebagai list kosong.

Didefinisikan `prime(n)` untuk mengecek apakah bilangan yang dimasukkan merupakan bilangan prima atau bukan. Kemudian, dengan penggunaan pengulangan `for` dilakukan untuk mencari seluruh kombinasi dari `n`.

`Set` digunakan untuk menyimpan nilai-nilai hasil kombinasi dari `n` dalam variabel `num`. Penggunaan `set` juga dimaksudkan agar output yang dihasilkan tidak memiliki duplikat.

Hasil seluruh kombinasi dari `n`  dalam variabel num akan dicek kembali apakah merupakan bilangan prima atau bukan dengan menggunakan pengulangan `for` sehingga nantinya output yang dihasilkan hanya list dari bilangan prima.

**Deskripsi Singkat Soal** <br>
$$\begin{bmatrix} S & 2 & 7 & 3 & 5 \\ 0 & 3 & 2 &4 & 2 \\ 5 & 1 & 8 & 7 & 0 \\ 6 & 2 & 1 & 9 & 3 \\ 4 & 7 & 6 & 0 & X \end{bmatrix}$$ <br>

Membuat program lintasan (penelusuran matriks) yang dapat membantu untuk menemukan harta karun $X$ dengan titik awal kamu pada $S$. Lintasan tersebut harus dengan *cost* yang paling minimum. Cara penelusuran tidak boleh diagonal. Cara penelusuran hanya boleh berjalan lurus ke bawah, kanan, atas, atau kiri. Output yang diharapkan dari program ini merupakan sebuah list yang mengandung upah yang telah dibayar.

```python
def mincostsum(B):
    m = len(B)
    n = len(B[0])
    A = [0]*m
    for i in range(m):
        A[i] = [0]*n
    A[0][1] = B[0][1]
    A[1][0] = B[1][0]
    row = len(B)
    column = len(B[0])
    for i in range(2,column):
        A[0][i] = B[0][i] + A[0][i-1]
    for i in range(2,row):
        A[i][0] = B[i][0] + A[i-1][0]
    for i in range(1,row):
        for j in range(1,column):
            if B[i][j]=="X":
                A[i][j] = 0 + min(A[i-1][j],A[i][j-1])
                break
            A[i][j] = B[i][j] + min(A[i-1][j],A[i][j-1])
    cost = list(A,B)
    return(A[i][j],cost)
def min(x,y):
    if x>y:
        min = y
    else:
        min = x
    return min
def list(A,B):
    row = len(B)
    column = len(B[0])
    list = []
    i = len(B) - 1
    j = len(B) - 1
    for x in range(1,row+column-2):
        if i<0:
            for y in range(j,1,-1):
                a = B[0][y]
                list.insert(0,a)
            break
        elif j<0:
            for z in range(i,1,-1):
                b = B[z][0]
                list.insert(0,b)
            break
        if A[i][j-1]==min(A[i][j-1],A[i-1][j]):
            c = B[i][j-1]
            list.insert(0,c)
            j = j - 1
        elif A[i-1][j]==min(A[i][j-1],A[i-1][j]):
            d = B[i-1][j]
            list.insert(0,d)
            i = i - 1
    return list
import sympy as sp
X = [["S", 2, 7, 3, 5], [0, 3, 2, 4, 2], [5, 1, 8, 7, 0], [6, 2, 1, 9, 3], [4, 7, 6, 0, "X"]]
print("====\tFind The Treasure!\t====\n")
Y = sp.Matrix(X)
sp.pprint(Y)
total, cost = mincostsum(X)
print("\nList Minimum Cost =",cost)
print("Minimum Total Cost =",total)
print("Route : S > {} > {} > {} > {} > {} > {} > {} > X".format(cost[0],cost[1],cost[2],cost[3],cost[4],cost[5],cost[6]))
print("\n====\tCongratulation!\t\t====")
```

**Penjelasan Program** <br>
Dalam program ini, digunakan 3 fungsi yaitu sebagai berikut.
- `def mincostsum(B)` : membuat matriks penjumlahan *cost* (A) yang dilalui dalam matriks B.
- `def min(x,y)` : menentukan nilai minimum antara x dengan y.
- `def list(A,B)` : membandingkan dua elemen A dari ujung kanan bawah (dimulai dari `A[4][3]` dan `A[3][4]`), kemudian membuat list minimum *cost* dari lintasan penelusuran dalam B dengan menyesuaikan terhadap index nilai minimum elemen dari matrix A yang telah dibandingkan.
1. **`def mincostsum(B)`** <br>
Karena cara penelusuran hanya boleh berjalan lurus ke bawah, kanan, atas, atau kiri, maka akan digunakan lintasan ke kanan atau bawah saja agar meminimumkan total *cost*. <br>
Langkah - langkahnya :
- Membuat matriks A dengan ukuran yang sama dengan matriks B, dimana elemen A
$$A[i][j] = \left\{
\begin{array}{ll}
      B[i][j] & \mbox{,  untuk } [i][j] = [0][1] \mbox{ atau } [1][0]\\
      0 & \mbox{,  untuk } [i][j] \mbox{ yang lainnya}\\
\end{array}
\right.$$
- Membuat jumlahan *cost* baris pertama dari B
$$\begin{bmatrix} 0 & 2 & 9 & 12 & 17 \end{bmatrix}$$
- Membuat jumlahan *cost* kolom pertama dari B
$$\begin{bmatrix} 0 \\ 0 \\ 5 \\ 11 \\ 15 \end{bmatrix}$$
- `A[i][j] = B[i][j] + min(A[i-1][j],A[i][j-1])` <br>
Syntax tersebut berfungsi untuk menghitung elemen `A[i][j]` dengan menjumlahkan `B[i][j]` dengan nilai minimum antara `A[i-1][j]` dan `A[i][j-1]`.
$$A = \begin{bmatrix} 0 & 2 & 9 & 12 & 17 \\ 0 & 3 & 5 & 9 & 11 \\ 5 & 4 & 12 & 16 & 11 \\ 11 & 6 & 7 & 16 & 14 \\ 15 & 13 & 13 & 13 & 13 \end{bmatrix}$$ <br>
- **Catatan** : <br>
Jika `B[i][j] = "X"`, akan diganti menjadi `B[i][j] = 0`.
2. **`def min(x,y)`** <br>
Jika `x > y`, maka nilai minimumnya yaitu y. Jika yang lainnya (`x <= y`), maka nilai minimumnya yaitu x.
3. **`def list(A,B)`** <br>
Fungsi ini berfungsi untuk membandingkan dua elemen A dari ujung kanan bawah (dimulai dari `A[4][3]` dan `A[3][4]`), dimana yang menjadi nilai minimumnya akan diambil indexnya kemudian dijadikan lintasan yang sesuai dengan index di B.
$$A = \begin{bmatrix} 0 & 2 & 9 & 12 & 17 \\ *0* & *3* & 5 & 9 & 11 \\ 5 & *4* & 12 & 16 & 11 \\ 11 & *6* & *7* & 16 & 14 \\ 15 & 13 & *13* & *13* & 13 \end{bmatrix}$$ <br>
$$B = \begin{bmatrix} *S* & 2 & 7 & 3 & 5 \\ *0* & *3* & 2 & 4 & 2 \\ 5 & *1* & 8 & 7 & 0 \\ 6 & *2* & *1* & 9 & 3 \\ 4 & 7 & *6* & *0* & *X* \end{bmatrix}$$ <br>
- **Catatan** : <br>
 - Jika `i < 0` atau `j < 0`, maka program akan error karena index matriks dimulai dari 0. Oleh karena itu, dibuat kondisi
   - `i < 0` (index barisnya negatif) <br>
   sisa lintasannya akan dilanjutkan dengan melewati baris `i + 1` ke kiri sampai `A[0][1]`
   - `j < 0` (index kolomnya negatif) <br>
   sisa lintasan akan dilanjutkan dengan melewati kolom `j + 1` ke atas sampai `A[1][0]`.
 - Tanda bintang pada A menandakan nilai minimum hasil perbandingan. <br>
 - Tanda bintang pada B menandakan jalur lintasan yang harus dilalui. <br>
